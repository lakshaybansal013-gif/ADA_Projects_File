```python
import pandas as pd
import matplotlib.pyplot as plt
```

#### Insertion Sort

```python
df=pd.read_csv('insertion_sort.csv')
df
```

```python
df.plot(x='size of array', y='Execution Time', kind='line', marker='o')
plt.xlabel('Size of Array')
plt.ylabel('Execution Time (seconds)')
plt.title('Insertion Sort Time Complexity')
plt.grid(True)
plt.show()
```
