```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('searching_time_complexity_l.csv')
df2=pd.read_csv('searching_time_complexity_b.csv')
print(df)
print(df2)
```

```python
plt.figure(figsize=(8, 4))
ax = plt.gca()

df.plot(
    x='n',
    y='Linear Search(sec)',
    kind='line',
    ax=ax,
    label='Linear Search',
    title='Search Time Complexity'
)

df2.plot(
    x='n',
    y='Binary Search(sec)',
    kind='line',
    ax=ax,
    label='Binary Search'
)

plt.xlabel('Input Size')
plt.ylabel('Time Complexity')
plt.legend()
plt.grid(True)
plt.show()
```
