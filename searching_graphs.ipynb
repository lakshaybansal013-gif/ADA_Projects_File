```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('searching_time_complexity_l.csv')
print(df)
```

```python
df2=pd.read_csv('searching_time_complexity_b.csv')
print(df2)
```

```python
df.plot(x='n', y='Linear Search(sec)',kind='line', marker='o')
plt.xlabel('Size of Array')
plt.ylabel('Execution Time (seconds)')
plt.title('Linear Search Time Complexity')
plt.grid(True)
plt.show()
```

```python
df2.plot(x='n', y='Binary Search(sec)',kind='line', marker='o')
plt.xlabel('Size of Array')
plt.ylabel('Execution Time (seconds)')
plt.title('Binary Search Time Complexity')
plt.grid(True)
plt.show()
```
